```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("iris.csv")

print(data.head())
```

```python
print("Columns in dataset:")
print(data.columns.tolist())

print("\nShape of dataset:", data.shape)
```

```python
target_column = "flower"

features = [
    "sepal length (cm)",
    "sepal width (cm)",
    "petal length (cm)",
    "petal width (cm)"
]

print("Features:")
print(features)

print("\nTarget column:")
print(target_column)
```

```python
print("Classes:")
print(data[target_column].unique())
```

```python
print("Missing values:")
print(data.isnull().sum())
```

```python
def learn_rule(data, features, target_column, target_class):

    best_rule = None
    best_score = -1

    for feature in features:

        values = sorted(data[feature].unique())

        for threshold in values:

            # Rule 1: feature <= threshold
            covered = data[data[feature] <= threshold]

            if len(covered) > 0:

                positive = len(
                    covered[covered[target_column] == target_class]
                )

                score = positive / len(covered)

                if score > best_score:
                    best_score = score
                    best_rule = (feature, "<=", threshold)

            # Rule 2: feature > threshold
            covered = data[data[feature] > threshold]

            if len(covered) > 0:

                positive = len(
                    covered[covered[target_column] == target_class]
                )

                score = positive / len(covered)

                if score > best_score:
                    best_score = score
                    best_rule = (feature, ">", threshold)

    return best_rule, best_score
```

```python
def apply_rule(data, rule):

    feature, operator, threshold = rule

    if operator == "<=":
        return data[data[feature] <= threshold]

    elif operator == ">":
        return data[data[feature] > threshold]
```

```python
def sequential_covering(data, features, target_column):

    rules = []

    classes = data[target_column].unique()

    for target_class in classes:

        remaining = data.copy()

        print("\nLearning rules for:", target_class)

        while len(
            remaining[remaining[target_column] == target_class]
        ) > 0:

            rule, score = learn_rule(
                remaining,
                features,
                target_column,
                target_class
            )

            if rule is None:
                break

            covered = apply_rule(remaining, rule)

            positive_covered = covered[
                covered[target_column] == target_class
            ]

            if len(positive_covered) == 0:
                break

            rules.append(
                (target_class, rule, score)
            )

            feature, operator, threshold = rule

            print(
                "IF", feature,
                operator, threshold,
                "THEN", target_column,
                "=", target_class,
                "| Score =", round(score, 2)
            )

            # Remove positive examples covered by the rule
            remaining = remaining.drop(
                positive_covered.index
            )

    return rules
```

```python
rules = sequential_covering(
    data,
    features,
    target_column
)
```

```python
print("========================================")
print("           FINAL RULES")
print("========================================")

for i, (target_class, rule, score) in enumerate(rules, 1):

    feature, operator, threshold = rule

    print(
        f"Rule {i}: "
        f"IF {feature} {operator} {threshold} "
        f"THEN {target_column} = {target_class}"
    )
```
